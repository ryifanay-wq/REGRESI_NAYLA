```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

#tugas 1
data = np.array([[1,2,3,4],[5,6,7,8]])

print (data)

#tugas 2
# T2.2
a = np.array([[1,2,3,4],[5,66,7,8]])
print(a.shape)

print(a)
print(type(a))
# T2.4
a = np.array([[1,2,3],[4,5,6]])
b = a.reshape(3,2)
print(b)
# T2.5
a = np.arange(24)
b = a.reshape(2,4,3)
print(b)

#tugas 3
# T3.3
print(np.arange(2, 6, 5))
# T3.4
print(np.arange(3, 10, 3))

#tugas 4

# T4.2
urutan = np.linspace(0, 10, 4, dtype=int)
print(urutan)

# Membuat array data
data = np.array([4, 9, 16, 25, 36])

# Operasi statistika
mean = np.mean(data)       # Rata-rata
median = np.median(data)   # Median
std_dev = np.std(data)     # Standar deviasi
variance = np.var(data)    # Variansi
min_value = np.min(data)   # Nilai minimum
max_value = np.max(data)   # Nilai maksimum

print("Rata-rata:", mean)
print("Median:", median)
print("Standar Deviasi:", std_dev)
print("Variansi:", variance)
print("Nilai Minimum:", min_value)
print("Nilai Maksimum:", max_value)

# Menghitung korelasi antara dua set data
data1 = np.array([12, 34, 56, 78, 90])
data2 = np.array([23, 45, 67, 89, 101])
correlation = np.corrcoef(data1, data2)[0, 1]
print("Korelasi antara data1 dan data2:", correlation)


# Membuat deret waktu dari 0 hingga 2*pi dengan 1000 titik
t = np.linspace(0, 2 * np.pi, 1000)

# Menghitung nilai sine untuk setiap titik waktu
y = np.sin(t)

# Membuat plot gelombang sinus
plt.figure(figsize=(8, 4))  # Mengatur ukuran gambar
plt.plot(t, y, label='Sine Wave', color='blue')  # Plot sine wave dengan label
plt.title('Plot Gelombang Sinus')  # Judul plot
plt.xlabel('Waktu (s)')  # Label sumbu x
plt.ylabel('Amplitudo')  # Label sumbu y
plt.grid(True)  # Menampilkan grid
plt.legend()  # Menampilkan legenda
plt.show()  # Menampilkan plot

# Membuat deret waktu dari 0 hingga 4*pi dengan 1000 titik

p = np.linspace(0, 4 * np.pi, 1000)


# Menghitung nilai sine dan cosine untuk setiap titik waktu

p_sin = np.sin(t)
p_cos = np.cos(t)


# Membuat plot gelombang sinus dan gelombang kosinus
plt.figure(figsize=(8, 4))
plt.plot(p, p_sin, label='Sine Wave', color='blue')
plt.plot(p, p_cos, label='Cosine Wave', color='red', linestyle='--')

plt.title('Plot Gelombang Sinus dan Gelombang Kosinus')
plt.xlabel('Waktu (s)')
plt.ylabel('Amplitudo')
plt.grid(True)
plt.legend()
plt.show()

# Membuat deret waktu dari 0 hingga 2*pi dengan 1000 titik

cx = np.linspace(0, 2 * np.pi, 1000)


# Menghitung nilai beberapa gelombang sinus dengan frekuensi yang berbeda
freqs = [1, 2, 3]  # Frekuensi gelombang sinus
amplitudes = [1, 0.7, 0.5]  # Amplitudo gelombang sinus
waves = []
for freq, amplitude in zip(freqs, amplitudes):

    wave = amplitude * np.sin(freq * cx)

    waves.append(wave)

# Menambahkan semua gelombang menjadi satu
composite_wave = np.sum(waves, axis=0)

# Membuat plot untuk semua gelombang dan gelombang komposit
plt.figure(figsize=(8, 4))

for i, (freq, amplitude) in enumerate(zip(freqs, amplitudes)):

    label = f'Sine Wave {i+1}'  # Label gelombang

    plt.plot(t, waves[i], label=label)
plt.plot(t, composite_wave, label='Composite Wave', color='black', linewidth=2)
plt.title('Plot Gelombang Sinus dengan Frekuensi Berbeda')
plt.xlabel('Waktu (s)')
plt.ylabel('Amplitudo')
plt.grid(True)
plt.legend()
plt.show()

# Parameters for the random walk
num_steps = 100
num_walks = 5


# Simulate random walks
walks = np.cumsum(np.random.choice([-1, 1], size=(num_steps, num_walks)), axis=0)


# Plot the random walks
plt.figure(figsize=(10, 6))
for i in range(num_walks):

    plt.plot(walks[:, i], label=f'Walk {i+1}')
plt.title('Random Walk Simulations')
plt.xlabel('Step')
plt.ylabel('Position')
plt.legend()
plt.grid()
plt.show()

#import data tabel pandas
df = pd.read_csv('https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/main/penjualan.csv')
print(df.head(5))

print(df.tail())
print(df)
d = {'tinggi': [320, 300], 'rendah': [95,100]}

df1 = pd.DataFrame(data=d)

print(df1)
# # Boleh menampilkan sesuai selera
# df.head(4) #> tampil 4 baris pertama
# df.tail()  #> tampil 5 baris terakhir

feb = df["feb"]

print(feb.head())

#data banyak
dfone = pd.read_csv('https://raw.githubusercontent.com/Alam-A99/DATAAKSI/main/dataset_fiskal_97_hipoFull.csv')
print(dfone)

print(df.loc[:, :])
print(df.loc[:,])
print(df.iloc[:, :])
df.loc[0:2, ['nama', 'feb', 'apr']]

df.iloc[0:4, [0, 1, 2]]

rata_rata = dfone.loc[:, "feb"].mean()
varians = dfone.loc[:, "feb"].var()
std_dev = dfone.loc[:, "feb"].std()

print("Rata-rata:", rata_rata)
print("Varians:", varians)
print("Standar deviasi:", std_dev)
```
