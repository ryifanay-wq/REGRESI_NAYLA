```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# =========================
# TUGAS 1: ARRAY NUMPY
# =========================
data = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print(data)


# =========================
# TUGAS 2: BENTUK ARRAY
# =========================
a = np.array([[1, 2, 3, 4], [5, 66, 7, 8]])
print(a.shape)
print(a)
print(type(a))

a = np.array([[1, 2, 3], [4, 5, 6]])
b = a.reshape(3, 2)
print(b)

a = np.arange(24)
b = a.reshape(2, 4, 3)
print(b)


# =========================
# TUGAS 3: ARANGE
# =========================
print(np.arange(2, 6, 5))
print(np.arange(3, 10, 3))


# =========================
# TUGAS 4: LINSPACE
# =========================
urutan = np.linspace(0, 10, 4, dtype=int)
print(urutan)


# =========================
# STATISTIKA NUMPY
# =========================
data = np.array([4, 9, 16, 25, 36])

print("Rata-rata:", np.mean(data))
print("Median:", np.median(data))
print("Standar Deviasi:", np.std(data))
print("Variansi:", np.var(data))
print("Nilai Minimum:", np.min(data))
print("Nilai Maksimum:", np.max(data))


# =========================
# KORELASI
# =========================
data1 = np.array([12, 34, 56, 78, 90])
data2 = np.array([23, 45, 67, 89, 101])

correlation = np.corrcoef(data1, data2)[0, 1]
print("Korelasi:", correlation)


# =========================
# GRAFIK GELOMBANG SINUS
# =========================
t = np.linspace(0, 2 * np.pi, 1000)
y = np.sin(t)

plt.figure(figsize=(8, 4))
plt.plot(t, y, label="Sine Wave")
plt.title("Plot Gelombang Sinus")
plt.xlabel("Waktu (s)")
plt.ylabel("Amplitudo")
plt.grid(True)
plt.legend()
plt.show()


# =========================
# GRAFIK SINUS DAN KOSINUS
# =========================
p = np.linspace(0, 4 * np.pi, 1000)
p_sin = np.sin(p)
p_cos = np.cos(p)

plt.figure(figsize=(8, 4))
plt.plot(p, p_sin, label="Sine Wave")
plt.plot(p, p_cos, label="Cosine Wave", linestyle="--")
plt.title("Plot Gelombang Sinus dan Kosinus")
plt.xlabel("Waktu (s)")
plt.ylabel("Amplitudo")
plt.grid(True)
plt.legend()
plt.show()


# =========================
# GELOMBANG DENGAN FREKUENSI BERBEDA
# =========================
cx = np.linspace(0, 2 * np.pi, 1000)
freqs = [1, 2, 3]
amplitudes = [1, 0.7, 0.5]
waves = []

for freq, amplitude in zip(freqs, amplitudes):
    wave = amplitude * np.sin(freq * cx)
    waves.append(wave)

composite_wave = np.sum(waves, axis=0)

plt.figure(figsize=(8, 4))

for i, wave in enumerate(waves):
    plt.plot(cx, wave, label=f"Sine Wave {i + 1}")

plt.plot(cx, composite_wave, label="Composite Wave",
         color="black", linewidth=2)
plt.title("Gelombang Sinus dengan Frekuensi Berbeda")
plt.xlabel("Waktu (s)")
plt.ylabel("Amplitudo")
plt.grid(True)
plt.legend()
plt.show()


# =========================
# RANDOM WALK
# =========================
num_steps = 100
num_walks = 5

walks = np.cumsum(
    np.random.choice([-1, 1], size=(num_steps, num_walks)),
    axis=0
)

plt.figure(figsize=(10, 6))

for i in range(num_walks):
    plt.plot(walks[:, i], label=f"Walk {i + 1}")

plt.title("Random Walk Simulations")
plt.xlabel("Step")
plt.ylabel("Position")
plt.legend()
plt.grid(True)
plt.show()


# =========================
# FUNGSI BANTU: CARI KOLOM
# =========================
def cari_kolom(dataframe, kata_kunci):
    """Cari nama kolom asli yang mengandung kata_kunci (tidak peduli
    huruf besar/kecil dan spasi). Mengembalikan None jika tidak ada."""
    for nama in dataframe.columns:
        if kata_kunci.lower() in str(nama).strip().lower():
            return nama
    return None


# =========================
# PANDAS: MEMBACA DATA PENJUALAN
# =========================
url = "https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/main/penjualan.csv"
df = pd.read_csv(url)

print(df.head(5))
print(df.tail())
print(df)

# Membuat DataFrame
d = {"tinggi": [320, 300], "rendah": [95, 100]}
df1 = pd.DataFrame(data=d)
print(df1)

# Menampilkan kolom Februari (nama kolom dicari otomatis)
print("Kolom tersedia:", df.columns.tolist())

kolom_feb = cari_kolom(df, "feb")
if kolom_feb is not None:
    print(f"Kolom Februari: '{kolom_feb}'")
    print(df[kolom_feb].head())
else:
    print("Kolom Februari tidak ditemukan. Pilih salah satu dari daftar di atas.")


# =========================
# PANDAS: DATASET KEDUA
# =========================
url2 = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/main/dataset_fiskal_97_hipoFull.csv"
dfone = pd.read_csv(url2)

print(dfone)
print("Kolom dataset kedua:", dfone.columns.tolist())


# =========================
# MENGAKSES DATA DENGAN LOC DAN ILOC
# =========================
print(df.loc[:, :])
print(df.iloc[:, :])

print(df.loc[0:2, :].head())
print(df.iloc[0:4, [0, 1, 2]])


# =========================
# STATISTIKA DATASET KEDUA
# =========================
kolom_feb2 = cari_kolom(dfone, "feb")

if kolom_feb2 is not None:
    seri = dfone[kolom_feb2]
    print(f"Statistik kolom '{kolom_feb2}':")
else:
    # Dataset kedua tidak punya kolom Februari -> pakai kolom numerik pertama
    kolom_numerik = dfone.select_dtypes(include="number").columns
    if len(kolom_numerik) == 0:
        raise ValueError("Dataset kedua tidak punya kolom numerik.")
    kolom_pakai = kolom_numerik[0]
    seri = dfone[kolom_pakai]
    print(f"Kolom Februari tidak ada di dataset kedua. "
          f"Statistik dihitung dari kolom '{kolom_pakai}':")

print("Rata-rata:", seri.mean())
print("Varians:", seri.var())
print("Standar deviasi:", seri.std())
```
